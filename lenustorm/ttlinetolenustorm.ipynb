## Dipole parameters

- units in [m], [rad], [GeV]

Can be initiated with:
    o K0, S-length

```python
# -- Defined dipole from K0 and physical length
dipole = dd()
dipole.k0 = 0.428
dipole.length = 2.21
dipole.angle = dipole.k0*dipole.length
dipole.angle_deg = dipole.angle*180/np.pi
dipole.deflection = dipole.length * np.tan(dipole.angle/2)
dipole.x = dipole.deflection/np.tan(dipole.angle)
dipole.pivot = dipole.length - dipole.x
dipole.pivot_from_center = dipole.pivot - dipole.length/2
dipole.bending_radius = dipole.pivot/np.tan(dipole.angle/2)
dipole.sagitta = dipole.bending_radius*(1-np.cos(dipole.angle/2))
dipole.slength = dipole.angle*dipole.bending_radius
dipole.field = 1.0
pprint.pprint(f' -- dipole : {dipole}')
```

```python
vars(dipole)
```

```python
print(f' ---- Momentum for 1.0T magnetic field ...{299.79*dipole.field*dipole.slength/(dipole.angle*1000)}')
print(f' ---- Field for 0.7GeV Beam momentum......{(dipole.angle*1000)*0.7/299.79/dipole.slength}')
```

```python
# --- Define dipole form angle and s-length
dipole2 = dd()
dipole2.k0 = 0.428
dipole2.slength = 2.21
dipole2.angle = dipole2.k0*dipole2.slength
dipole2.angle_deg = dipole2.angle*180/np.pi
dipole2.bending_radius = dipole2.slength/dipole2.angle
dipole2.pivot = dipole2.bending_radius*np.tan(dipole2.angle/2)
dipole2.x = 2*dipole2.bending_radius*np.sin(dipole2.angle/2)
dipole2.length = dipole2.x*np.cos(dipole2.angle/2)
dipole2.deflection = dipole2.length * np.tan(dipole2.angle/2)
dipole2.pivot_from_center = dipole2.pivot - dipole2.length/2
dipole2.sagitta = dipole2.bending_radius*(1-np.cos(dipole2.angle/2))
dipole2.field = 1.0
pprint.pprint(f' -- dipole : {dipole2}')
```
